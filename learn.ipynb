```ruby
def calculate_progress(total_tasks, completed_tasks)
  return 0 if total_tasks == 0

  completed_tasks.to_f / total_tasks * 100
end

progress = calculate_progress(10, 7)

puts "Progress: #{progress}%"
```

```ruby
def status_message(status)
  case status
  when "pending"
    "Task has not started"
  when "in_progress"
    "Task is currently running"
  when "completed"
    "Task completed"
  else
    "Invalid status"
  end
end

puts status_message("pending")
```
